```python
from rdkit import Chem
from rdkit.Chem import Draw, rdDepictor

def draw(smiles, size=(900, 600)):
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        print("Neplatné SMILES")
        return
    return Draw.MolToImage(mol, size=size)

rdDepictor.SetPreferCoordGen(True)
smile = 'CC=C(C)C(=O)OC1C(OC(C)=O)C2(CO)C(O)CC3(C)C(=CCC4C5(C)CCC(OC6OC(C(=O)O)C(OC7OC(CO)C(O)C(O)C7O)C(O)C6OC6OC(CO)C(O)C(O)C6O)C(C)(CO)C5CCC43C)C2CC1(C)C'
draw(smile, size = (500,300))
```
